```python
import os, glob, subprocess
from pathlib import Path

# ----------------------------
# PATHS (edit if needed)
# ----------------------------
DATASET_DIR = "/kaggle/working/data"
SUBJECTS_DIR = "/kaggle/working/data/ds003126_fastsurfer"
LICENSE_PATH = "/kaggle/input/license/license.txt"
FASTSURFER_BIN = "/kaggle/working/FastSurfer/run_fastsurfer.sh"
OUTPUT_DIR = "/kaggle/working/fastsurfer_seg_only_tars"
DELETE_AFTER_TAR = True

os.makedirs(SUBJECTS_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ----------------------------
# TD SUBJECT LIST (ONLY these)
# ----------------------------
TD_SUBJECTS = [
    "sub-047EPKL014007","sub-047EPKL014008","sub-047EPKL014010","sub-047EPKL014020",
    "sub-047EPKL014024","sub-047EPKL014026","sub-047EPKL014029","sub-047EPKL014031",
    "sub-047EPKL014036","sub-047EPKL014037","sub-047EPKL014038","sub-047EPKL014039",
    "sub-047EPKL014040","sub-047EPKL014043","sub-047EPKL014045","sub-047EPKL014048",
    "sub-047EPKL014050","sub-047EPKL014052","sub-047EPKL014054","sub-047EPKL014062",
    "sub-047EPKL014066","sub-047EPKL014067",
]

# ----------------------------
# FIND TD T1s ONLY (BIDS-correct)
# ----------------------------
t1_files = []
for sub in TD_SUBJECTS:
    # Match any session (ses-1, etc.) in case some differ
    pattern = os.path.join(DATASET_DIR, sub, "ses-*", "anat", f"{sub}_ses-*_T1w.nii.gz")
    matches = sorted(glob.glob(pattern))
    if len(matches) == 0:
        print(f"❌ No T1 found for {sub} with pattern:\n   {pattern}")
        continue
    # If multiple sessions match, we’ll run each session
    t1_files.extend(matches)

t1_files = sorted(set(t1_files))

print(f"\n✅ TD-only: Found {len(t1_files)} T1 files total across TD subjects.")
for f in t1_files[:10]:
    print("   -", f)

if len(t1_files) == 0:
    root_items = sorted([p.name for p in Path(DATASET_DIR).glob("*")])[:30]
    print("\n❌ No TD T1s matched. Top-level items in DATASET_DIR (first 30):")
    print(root_items)
    raise SystemExit("Fix DATASET_DIR or TD subject paths, then rerun.")

# ----------------------------
# Helper: create unique SID sub+session
# ----------------------------
def parse_sid_from_path(t1_path: str) -> str:
    p = Path(t1_path)
    ses = p.parent.parent.name      # ses-1
    sub = p.parent.parent.parent.name  # sub-XXXX
    return f"{sub}_{ses}"

# ----------------------------
# RUN FASTSURFER FOR EACH TD T1
# ----------------------------
ok, missing, failed, skipped = 0, 0, 0, 0

for t1_path in t1_files:
    sid = parse_sid_from_path(t1_path)
    t1_path = os.path.abspath(t1_path)

    output_tar = os.path.join(OUTPUT_DIR, f"{sid}_GPU_SEG_ONLY.tar.gz")
    subj_folder = os.path.join(SUBJECTS_DIR, sid)

    if not os.path.exists(t1_path):
        print(f"❌ Missing file on disk (unexpected): {t1_path}")
        missing += 1
        continue

    if os.path.exists(output_tar):
        print(f"⚠️ Exists, skipping: {output_tar}")
        skipped += 1
        continue

    print(f"\n🚀 >>> SEGMENTING (TD): {sid} <<< 🚀")
    print("T1:", t1_path)

    cmd = f"""
    set -e
    export FS_LICENSE="{LICENSE_PATH}"
    "{FASTSURFER_BIN}" \
        --sid "{sid}" \
        --sd "{SUBJECTS_DIR}" \
        --t1 "{t1_path}" \
        --seg_only \
        --device cuda \
        --allow_root
    """

    process = subprocess.run(cmd, shell=True, executable="/bin/bash")

    if process.returncode == 0:
        print(f"📦 Tarring results for {sid} → {output_tar}")
        subprocess.run(
            f'tar -czf "{output_tar}" -C "{SUBJECTS_DIR}" "{sid}"',
            shell=True,
            check=True
        )

        if DELETE_AFTER_TAR:
            subprocess.run(f'rm -rf "{subj_folder}"', shell=True, check=True)
            print(f"🧹 Deleted uncompressed folder: {subj_folder}")

        print(f"✅ {sid} complete.")
        ok += 1
    else:
        print(f"❌ Error processing {sid}")
        failed += 1

print("\n🎉 DONE: TD-only GPU seg-only")
print(f"✅ Success: {ok}")
print(f"⚠️ Skipped (already tarred): {skipped}")
print(f"❌ Failed: {failed}")
print(f"❌ Missing: {missing}")

print("\n📁 Output tar examples:")
for p in sorted(glob.glob(os.path.join(OUTPUT_DIR, "*_GPU_SEG_ONLY.tar.gz")))[:10]:
    print("   -", p)
```
