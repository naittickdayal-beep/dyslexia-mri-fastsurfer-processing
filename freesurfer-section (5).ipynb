```python
import os, re

wrapper_path = "/kaggle/working/FastSurfer/recon_surf/spherically_project_wrapper.py"
assert os.path.exists(wrapper_path), f"Missing: {wrapper_path}"

with open(wrapper_path, "r", encoding="utf-8") as f:
    s = f.read()

# If already patched, stop
if "FORCE_FREESURFER_QSPHERE_FALLBACK" in s:
    print("✅ Wrapper already patched for qsphere fallback.")
else:
    # We inject a post-fallback copy step right after the fallback recon-all finishes.
    # Find a good anchor: the code that runs the fallback command.
    # We'll append copy logic immediately after the fallback finishes successfully.
    #
    # This is a heuristic patch: it looks for "Running fallback command:" and inserts code after
    # the subprocess call that runs recon-all.
    #
    # If your wrapper differs, search for the subprocess call that runs recon-all -qsphere and insert there.

    # 1) Add helper snippet (copy qsphere to expected outputs)
    helper = r'''
# === FORCE_FREESURFER_QSPHERE_FALLBACK ===
def _copy_fs_qsphere_to_expected_outputs(hemi: str, subj: str, subjects_dir: str):
    """
    After FreeSurfer recon-all -qsphere succeeds, copy the qsphere output into the
    filenames FastSurfer expects so downstream steps can continue.
    """
    surf_dir = os.path.join(subjects_dir, subj, "surf")
    # Common FreeSurfer outputs
    candidates = [
        os.path.join(surf_dir, f"{hemi}.qsphere.nofix"),
        os.path.join(surf_dir, f"{hemi}.qsphere"),
        os.path.join(surf_dir, f"{hemi}.sphere"),  # sometimes already exists
    ]
    src = next((p for p in candidates if os.path.exists(p)), None)
    if src is None:
        raise FileNotFoundError(f"Could not find qsphere/sphere output for hemi={hemi} in {surf_dir}")

    # Targets that downstream may look for
    targets = [
        os.path.join(surf_dir, f"{hemi}.sphere"),
        os.path.join(surf_dir, f"{hemi}.sphere.reg"),
    ]

    for t in targets:
        try:
            # If already there and non-empty, leave it
            if os.path.exists(t) and os.path.getsize(t) > 0:
                continue
            import shutil
            shutil.copy(src, t)
        except Exception as e:
            print(f"WARNING: could not copy {src} -> {t}: {e}")
# === END FORCE_FREESURFER_QSPHERE_FALLBACK ===
'''

    # Inject helper near the top (after imports)
    if "def _copy_fs_qsphere_to_expected_outputs" not in s:
        s = re.sub(r"(\nimport .*?\n)", r"\1" + helper + "\n", s, count=1, flags=re.DOTALL)

    # 2) After fallback recon-all returns success, call the copier and exit 0
    # We patch the block that prints "Running fallback command:" and then runs the command.
    # We'll look for the line that assigns/gets the return code from that subprocess and inject.
    #
    # Common patterns include something like:
    #   process = PyPopen(...)
    #   done = process.finish(...)
    # We'll inject after that in the exception handler.

    insertion = r'''
            # If fallback succeeded, use FreeSurfer qsphere outputs and continue successfully.
            try:
                _copy_fs_qsphere_to_expected_outputs(args.hemi, args.subject, os.path.dirname(args.sdir))
            except Exception as _e:
                print(f"WARNING: qsphere copy failed: {_e}")
            sys.exit(0)
'''

    # Find the exception handler for the projection failure and insert right after fallback finishes.
    # Heuristic: insert after the string "recon-all -s" appears in code AND after a finish()/wait.
    if "sys.exit(0)" not in s or "qsphere copy failed" not in s:
        # Insert after a line containing "Running fallback command"
        s = re.sub(
            r"(print\(.*Running fallback command:.*\n(?:.*\n){0,40}?.*finish\(.*\)\n)",
            r"\1" + insertion,
            s,
            count=1
        )

    with open(wrapper_path, "w", encoding="utf-8") as f:
        f.write(s)

    print("✅ Patched wrapper to accept FreeSurfer qsphere fallback and exit success.")
```

```python
import os, re, shutil, subprocess, glob, multiprocessing

# ============================================================
# SURF-ONLY PIPELINE (BATCH SLICE) + FastSurfer NumPy2 FIX
# Change SUBJECT_START to move to the next chunk of subjects.
# ============================================================

# ---------- CONFIG ----------
GPU_DS = "/kaggle/input/datasets/naittickdayal/gpu-seg-control-children"
SUBJECTS_DIR = "/kaggle/working/subjects"
FS_LICENSE = "/kaggle/working/license.txt"
FREESURFER_HOME = "/kaggle/working/freesurfer"
FASTSURFER_HOME = "/kaggle/working/FastSurfer"
FASTSURFER_BIN = os.path.join(FASTSURFER_HOME, "run_fastsurfer.sh")

# ✅ Run "next 5":
SUBJECT_START = 20     # ✅ 0 = first five, 5 = next five, 10 = next five, 15 = next five...
MAX_SUBJECTS = 3       # chunk size

CPU_CORES = multiprocessing.cpu_count()
SKIP_IF_ZIP_EXISTS = True
DELETE_SUBJECT_AFTER_ZIP = False

os.makedirs(SUBJECTS_DIR, exist_ok=True)

# ---------- sanity checks ----------
assert os.path.exists(GPU_DS), f"Missing GPU_DS: {GPU_DS}"
assert os.path.exists(FASTSURFER_BIN), f"Missing FastSurfer script: {FASTSURFER_BIN}"
assert os.path.exists(FREESURFER_HOME), f"Missing FreeSurfer home: {FREESURFER_HOME}"
assert os.path.exists(FS_LICENSE), f"Missing FS license: {FS_LICENSE}"
assert os.path.exists(os.path.join(FREESURFER_HOME, "SetUpFreeSurfer.sh")), "Missing SetUpFreeSurfer.sh"

FS_PYTHON = os.path.join(FREESURFER_HOME, "python", "bin", "python3")
if not os.path.exists(FS_PYTHON):
    raise FileNotFoundError(f"FS_PYTHON not found at {FS_PYTHON}.")

# ============================================================
# 0) PATCH FASTSURFER BUG FOR NUMPY 2.x (newbyteorder removed)
# ============================================================
def patch_fastsurfer_numpy2():
    target = os.path.join(FASTSURFER_HOME, "recon_surf", "image_io.py")
    if not os.path.exists(target):
        print(f"⚠️ Could not find {target} to patch. Skipping patch.")
        return

    with open(target, "r", encoding="utf-8", errors="ignore") as f:
        txt = f.read()

    old = "data = data.byteswap().newbyteorder()"
    new = "data = data.byteswap().view(data.dtype.newbyteorder())"

    if old in txt:
        with open(target, "w", encoding="utf-8") as f:
            f.write(txt.replace(old, new))
        print("✅ Patched FastSurfer for NumPy 2.x in recon_surf/image_io.py")
    else:
        print("ℹ️ FastSurfer NumPy2 patch not needed (line not found).")

patch_fastsurfer_numpy2()

# ---------- discover input folders ----------
all_input_folders = sorted([
    f for f in os.listdir(GPU_DS)
    if os.path.isdir(os.path.join(GPU_DS, f))
])

print(f"📂 GPU_DS: {GPU_DS}")
print(f"✅ Found {len(all_input_folders)} folders total.")

# ✅ Slice to "next 5"
chunk = all_input_folders[SUBJECT_START:SUBJECT_START + MAX_SUBJECTS]
print(f"➡️ Will process folders [{SUBJECT_START}:{SUBJECT_START + MAX_SUBJECTS}] "
      f"= {len(chunk)} subjects")

def extract_sid(folder_name: str):
    m = re.search(r"(sub-[A-Za-z0-9]+)", folder_name)
    return m.group(1) if m else None

def first_or_none(lst):
    return lst[0] if lst else None

ok, failed, skipped, missing_inputs = 0, 0, 0, 0

for folder_name in chunk:
    sid = extract_sid(folder_name)
    if sid is None:
        print(f"⚠️ Skipping folder (no sid found): {folder_name}")
        skipped += 1
        continue

    src = os.path.join(GPU_DS, folder_name)
    out_subject = os.path.join(SUBJECTS_DIR, sid)
    done_marker = os.path.join(out_subject, "surf", "lh.pial")
    zip_path = f"/kaggle/working/{sid}_SURF_ONLY.zip"

    if SKIP_IF_ZIP_EXISTS and os.path.exists(zip_path):
        print(f"⚠️ Zip exists, skipping: {sid} → {zip_path}")
        skipped += 1
        continue

    if os.path.exists(done_marker):
        if not os.path.exists(zip_path):
            shutil.make_archive(f"/kaggle/working/{sid}_SURF_ONLY", "zip", out_subject)
        print(f"✅ {sid} already completed. Zip ready: {zip_path}")
        ok += 1
        continue

    print(f"\n{'='*70}\n🚀 PROCESSING: {sid}   (source: {folder_name})\n{'='*70}")

    # Clean output
    if os.path.exists(out_subject):
        shutil.rmtree(out_subject)

    mri_dir = os.path.join(out_subject, "mri")
    os.makedirs(mri_dir, exist_ok=True)

    # --- Locate required files under src ---
    orig_src = first_or_none(glob.glob(os.path.join(src, "**", "orig.mgz"), recursive=True))
    seg_src  = first_or_none(glob.glob(os.path.join(src, "**", "aparc.DKTatlas+aseg.deep.mgz"), recursive=True))
    mask_src = first_or_none(glob.glob(os.path.join(src, "**", "mask.mgz"), recursive=True))

    aseg_auto_src      = first_or_none(glob.glob(os.path.join(src, "**", "aseg.auto.mgz"), recursive=True))
    aseg_auto_noCC_src = first_or_none(glob.glob(os.path.join(src, "**", "aseg.auto_noCCseg.mgz"), recursive=True))

    if orig_src is None:
        print(f"❌ {sid}: missing orig.mgz")
        missing_inputs += 1; failed += 1; continue
    if seg_src is None:
        print(f"❌ {sid}: missing aparc.DKTatlas+aseg.deep.mgz")
        missing_inputs += 1; failed += 1; continue
    if mask_src is None:
        print(f"❌ {sid}: missing mask.mgz")
        missing_inputs += 1; failed += 1; continue

    # --- Copy into SUBJECTS_DIR ---
    orig_dst = os.path.join(mri_dir, "orig.mgz")
    seg_dst  = os.path.join(mri_dir, "aparc.DKTatlas+aseg.deep.mgz")
    mask_dst = os.path.join(mri_dir, "mask.mgz")
    aseg_dst = os.path.join(mri_dir, "aseg.auto.mgz")

    shutil.copy(orig_src, orig_dst)
    shutil.copy(seg_src, seg_dst)
    shutil.copy(mask_src, mask_dst)

    if aseg_auto_src:
        shutil.copy(aseg_auto_src, aseg_dst)
        aseg_note = "copied aseg.auto.mgz"
    elif aseg_auto_noCC_src:
        shutil.copy(aseg_auto_noCC_src, aseg_dst)
        aseg_note = "created aseg.auto.mgz from aseg.auto_noCCseg.mgz"
    else:
        print(f"❌ {sid}: missing aseg.auto.mgz and aseg.auto_noCCseg.mgz")
        missing_inputs += 1; failed += 1; continue

    print("✅ Inputs prepared:")
    print("   orig:", orig_dst)
    print("   seg :", seg_dst)
    print("   mask:", mask_dst)
    print("   aseg:", aseg_dst, f"({aseg_note})")

    # --- Run FastSurfer surf-only ---
    cmd = f"""
set -e
export FREESURFER_HOME="{FREESURFER_HOME}"
export SUBJECTS_DIR="{SUBJECTS_DIR}"
export FS_LICENSE="{FS_LICENSE}"
export FS_PYTHON="{FS_PYTHON}"
export FASTSURFER_HOME="{FASTSURFER_HOME}"

source "{FREESURFER_HOME}/SetUpFreeSurfer.sh"

"{FASTSURFER_BIN}" \
  --sid "{sid}" \
  --sd "{SUBJECTS_DIR}" \
  --t1 "{orig_dst}" \
  --asegdkt_segfile "{seg_dst}" \
  --surf_only \
  --device cpu \
  --threads {min(CPU_CORES, 8)} \
  --allow_root
"""
    p = subprocess.run(cmd, shell=True, executable="/bin/bash")

    if os.path.exists(done_marker):
        print(f"🏁 SUCCESS: {sid}")

        shutil.make_archive(f"/kaggle/working/{sid}_SURF_ONLY", "zip", out_subject)
        print(f"📦 Zip created: {zip_path}")

        if DELETE_SUBJECT_AFTER_ZIP:
            shutil.rmtree(out_subject, ignore_errors=True)
            print(f"🧹 Deleted subject folder: {out_subject}")
        else:
            shutil.rmtree(mri_dir, ignore_errors=True)
            shutil.rmtree(os.path.join(out_subject, "scripts"), ignore_errors=True)
            print(f"🧹 Purged heavy folders for {sid} (kept surf/stats).")

        ok += 1
    else:
        print(f"❌ FAILED: {sid} (return code {p.returncode})")
        build_log = os.path.join(out_subject, "scripts", "build.log")
        if os.path.exists(build_log):
            try:
                with open(build_log, "r", errors="ignore") as f:
                    tail = f.readlines()[-60:]
                print("----- tail scripts/build.log -----")
                print("".join(tail))
                print("----------------------------------")
            except Exception as e:
                print(f"(Could not read build.log: {e})")
        failed += 1

print("\n==================== SUMMARY ====================")
print(f"✅ Success: {ok}")
print(f"⚠️ Skipped: {skipped}")
print(f"❌ Failed:  {failed}")
print(f"📄 Missing inputs: {missing_inputs}")
print("=================================================")
```
